## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My project is a **Binary Classification** task. The goal is to predict whether a content page is likely to decline in the future or not based on its historical search and engagement signals. After predicting the probability of decline, the pages will be ranked by risk so that reviewers can inspect the highest-risk pages first. I chose classification because the primary decision is whether a page is at risk of future decline, while the ranking is used to prioritize review.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

My final target is **future content decline**. I want to predict whether a content page is likely to lose performance in a future time window based on its historical search and engagement signals.

For the starter dataset, I will use **trend_direction == "down"** as a proxy label because the dataset does not contain future outcomes. In the full warehouse dataset, I plan to define the target using a future time window (for example, previous 90 days of features to predict decline in the next 30 days). Therefore, the starter label is a defined proxy, while the final capstone target will be an observed future outcome.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

My main success metric will be **Precision@50** because content reviewers can only inspect a limited number of pages. A good model should place the most important pages near the top of the ranked list. I will consider the model successful if it improves Precision@50 compared with a simple baseline rule, meaning that more of the top 50 recommended pages are truly at risk of future decline.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can only look at one or two conditions at a time, such as "CTR < 0.5" or "impressions > 500". However, future content decline depends on the combination of many factors, including impressions, clicks, CTR, average position, content age, engagement, and trend history. Machine learning can learn these complex patterns from historical data and estimate the probability of future decline. This helps reviewers prioritize pages more accurately than relying on a few hand-written rules.
